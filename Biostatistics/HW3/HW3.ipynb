Example 1.

The sample includes 60 observations from the cdc dataset. Their mean weight is 173.3 pounds, and the sample standard deviation is 49.04  pounds. Calculate an approximate 95 percent confidence interval for the mean weight of adults in the United States.

```python
import math
from scipy.stats import t as t_distribution
import numpy as np
from scipy import stats
```

```python
from math import sqrt

# sample data
mean = 173.3
sd = 49.04
n = 60

# standard error
se = sd / sqrt(n)

# z* = 1.96

# CI
lower = mean - 1.96 * se
upper = mean + 1.96 * se

print(f"Standard error: {se:.2f} lbs")
print(f"We are 95% confident that the mean weight of adults in the United States is between ({lower:.2f} and {upper:.2f}) lbs")
```

**Test: two-tailed one-sample t-test.**

Under standard conditions, the mean protein concentration in laboratory mice is 20 mg/mL. A researcher hypothesizes that a new diet changes this concentration.




```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
```

```python
# 1. Input data: protein concentrations measured in 9 mice.
data = [18, 19, 20, 21, 22, 23, 24, 25, 26]


mu_0 = 20
# Significance level.
alpha = 0.05

print("1. HYPOTHESES")
print(f"H₀: the population mean concentration equals {mu_0} mg/mL.")
print(f"H₁: the population mean concentration differs from {mu_0} mg/mL.")
```

```python
# 2. Calculate the sample mean and sd.
n = len(data)
mean = np.mean(data)

sd = np.std(data, ddof=1)  #ddof=1 means that when computing the sample variance, we divide by n − 1.

# 3. Calculate the standard error of the mean.
se = sd / np.sqrt(n)

# t-statistics & degrees of freedome.
t_stat = (mean - mu_0) / se
df = n - 1

print(f"t_stat = {t_stat} - how far the sample mean deviates from 20, measured in units of the standard error.")
```

```python
# t_stat & p-value
t_crit = stats.t.ppf(1 - alpha / 2, df)
p_value = 2 * stats.t.sf(abs(t_stat), df) #0.975

print(f"t_crit = {t_crit}")
print(f"p_value = {p_value}")
```

```python
stats.ttest_1samp(data, mu_0)
```

```python
stats.ttest_1samp(data, 15, alternative="greater")   # H₁: μ > 15
stats.ttest_1samp(data, 15, alternative="less")      # H₁: μ < 15
```

```python
print("Reject H0" if p_value < alpha else "Fail to reject H0")
```

**Two-sample Student’s t-test for independent groups with equal variances**



```python
control = np.array([8, 9, 10, 11, 12])
experiment = np.array([11, 12, 13, 14, 15])

n1, n2 = len(control), len(experiment)

mean1, mean2 = np.mean(control), np.mean(experiment)
sd1, sd2 = np.std(control, ddof=1), np.std(experiment, ddof=1)

pooled_var = (
    (n1 - 1) * sd1**2 + (n2 - 1) * sd2**2
) / (n1 + n2 - 2)

# SE.
se = np.sqrt(pooled_var * (1 / n1 + 1 / n2))

mean_diff = mean1 - mean2
print(f"mean_diff = {mean_diff}")
```

```python
t_stat = mean_diff / se
df = n1 + n2 - 2

print(f"t_stat = {t_stat}")
```

```python
t_crit = stats.t.ppf(1 - alpha / 2, df)
p_value = 2 * stats.t.sf(abs(t_stat), df)

print(f"t_crit = {t_crit}")

print(f"p_value = {p_value}")
```

```python
# equal_var=True
result = stats.ttest_ind(control, experiment, equal_var=True)
print(result)
```

A researcher is studying how a new compound affects liver enzyme activity in laboratory mice. Enzyme activity (U/L) was measured in a control group of 5 mice and in an experimental group of 8 mice that received the compound.

```python
control = np.array([8, 9, 10, 11, 12])
experiment = np.array([9, 11, 13, 15, 17, 19, 21, 23])

n1, n2 = len(control), len(experiment)

mean1, mean2 = np.mean(control), np.mean(experiment)
sd1, sd2 = np.std(control, ddof=1), np.std(experiment, ddof=1)

a = sd1**2 / n1
b = sd2**2 / n2

se = np.sqrt(a + b)

mean_diff = mean1 - mean2
t_stat = mean_diff / se

df = (a + b)**2 / (
    a**2 / (n1 - 1) + b**2 / (n2 - 1)
)

t_crit = stats.t.ppf(1 - alpha / 2, df)
p_value = 2 * stats.t.sf(abs(t_stat), df)

print("Reject H0" if p_value < alpha else "Fail to reject H0")

# equal_var=False: Welch test.
result = stats.ttest_ind(control, experiment, equal_var=False)
print(result)
```

Scenario 1. Blood glucose

Blood glucose (mg/dL) was measured in 7 mice. The same mice were then given a new drug, and glucose was measured again. Does the drug change blood glucose?

```python
before = np.array([142, 155, 138, 160, 149, 151, 145])
after  = np.array([131, 147, 130, 148, 141, 144, 136])
#H0 - mean c(glucose) didnt change
#H1 - mean c(glucose) did change
#Since there ar paired observations - use paired t test

alpha = 0.05
diff = before - after
mu_diff = np.mean(diff)
sd_diff = np.std(diff, ddof = 1)
t = mu_diff * np.sqrt(len(diff)) / sd_diff
t_crit = stats.t.ppf(1 - alpha / 2, len(diff)-1)

if abs(t) >= t_crit:
    print('H0 rejected. there is significant difference')
else:
    print('H0 cannot be rejected')

```

Scenario 2. Spring and summer mice

A researcher measured the body weight (g) of 8 mice in March. In June, the lab received a new batch of 8 mice from the same supplier, and they were weighed too. Does mean body weight differ between the two batches?

```python
march = np.array([24.1, 25.3, 23.8, 26.0, 24.7, 25.5, 24.9, 23.6])
june  = np.array([26.2, 27.0, 25.4, 27.8, 26.5, 25.9, 27.3, 26.1])

#Since groups are unrelated - use independent t test
#H0 - mean body weight didn't change
#H1 - mean body weight changed
mean_march = np.mean(march)
std_march = np.std(march, ddof = 1)

mean_june = np.mean(june)
std_june = np.std(june, ddof = 1)

pooled_variance = np.sqrt(
    ((len(march) - 1) * (std_march**2) + (len(june) - 1) * (std_june**2)) 
    / (len(march) + len(june) - 2)
)


t = (mean_march - mean_june) / (pooled_variance * sqrt((1 / len(march)) + (1 / len(june))))


ddf = len(march) + len(june) - 2
alpha = 0.05

t_crit = stats.t.ppf(1 - alpha / 2, ddf)
if abs(t) >= t_crit:
    print('H0 rejected. there is significant difference')
else:
    print('H0 cannot be rejected')

print('manual test')
print(t)
t_stat, p_val = stats.ttest_ind(march, june, equal_var=True)

print('scipy results')
print(t_stat, p_val)
```

Scenario 3. Stress hormone

Cortisol levels (ng/mL) were measured in 6 control mice and in 10 different mice exposed to a stress protocol. Does the stress protocol change cortisol levels?

```python
control  = np.array([50, 52, 48, 51, 49, 53])
stressed = np.array([45, 60, 72, 38, 85, 55, 66, 90, 41, 70])

#Check if data homoscedantic
#H0 - homoscedantic
#H1 - heteroscedantic

stat, p_value = stats.levene(control, stressed, center='median')
print(f'Levene statistic is {stat}, p-value is {p_value}') 
#Data is heteroscedantic. Will perform Welch test for independent samples
#H0 - no changes
#H1 - there are changes

stat, p_value = stats.ttest_ind(control, stressed, equal_var=False)
print(f'Welch t statistic is {stat}, p-value is {p_value}')

#p-value > alpha = 0.05, there are no significatn changes
```

Scenario 4. Left and right kidney

In 6 mice, both the left and the right kidney were weighed (g). Is there a difference in weight between the left and right kidney?

```python
left  = np.array([1.82, 1.75, 1.90, 1.68, 1.79, 1.85])
right = np.array([1.78, 1.70, 1.87, 1.62, 1.74, 1.80])

stat, p_value = stats.levene(left, right, center='median')
print(f'Levene test value is {stat}(p = {p_value}). Data is homoscedantic')

#Statistics are paired --> use paired t test

t, p = stats.ttest_rel(left,right)

print(f'T statistic is {t} (p = {p})')

#H0 is rejected. There is a difference in weight between l and r kidneys
```

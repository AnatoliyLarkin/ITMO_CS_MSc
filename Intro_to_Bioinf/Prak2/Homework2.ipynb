## Number of reads before trimming (raw reads)


```python
%%bash

gunzip data/SRR19238653_1.fastq.gz
gunzip data/SRR19238653_2.fastq.gz
```

```python
%%bash

nr1=$(cat data/SRR19238653_1.fastq | wc -l)
nr2=$(cat data/SRR19238653_2.fastq | wc -l)

number_of_reads_raw=$(($nr1 + $nr2))

echo $number_of_reads_raw
```

## Read length (raw reads)


```python
%%bash

read_length=$(($(cat data/SRR19238653_1.fastq | head -n 2 | tail -n 1 | wc -m)-1))
echo $read_length
```

## What factors would make you decide to trim your reads before alignment?

1. The discribution of qualities of terminal nucleotides is mediocre with outliers below 20. Decided to trim it to avoid potential misalignment
2. On GC content plot there is a small peak on 4-6 %GC. Means there is higher percentage of AT-rich sequences that expected. I suggested it could be some junk reads or technical sequences that can be automatically detected and deduplicated/removed. Also it could be polyN sequences so i wanted to remove this as well.

As per qc plot after trimming, this peak wasnt removed so it probably linked to other processed. Nevertheless, I decided to proceed with trimming, as it, as least, removed 7% of duplicates.

## Percentage of duplicate reads in the BAM file (after marking duplicates)

```python
%%bash

gatk SelectVariants -V results/variants.vcf -select-type SNP -O results/variants_snps.vcf
gatk SelectVariants -V results/variants.vcf -select-type INDEL -O results/variants_indels.vcf
```

```python
%%bash
#snps filtration
gatk VariantFiltration \
    -V "results/variants_snps.vcf" \
    -filter "DP < 10.0" --filter-name "DP10" \
    -filter "QD < 2.0" --filter-name "QD2" \
    -filter "QUAL < 30.0" --filter-name "QUAL30" \
    -filter "SOR > 3.0" --filter-name "SOR3" \
    -filter "FS > 60.0" --filter-name "FS60" \
    -filter "MQ < 40.0" --filter-name "MQ40" \
    -O "results/variants_snps_filtered.vcf"

    #-filter "MQRankSum < -12.5" --filter-name "MQRankSum" \
    #-filter "ReadPosRankSum < -8.0" --filter-name "ReadPosRankSum" \
```

```python
%%bash
#indels filtration

gatk VariantFiltration \
    -V "results/variants_indels.vcf" \
    -filter "DP < 10.0" --filter-name "DP10" \
    -filter "QD < 2.0" --filter-name "QD2" \
    -filter "QUAL < 30.0" --filter-name "QUAL30" \
    -filter "FS > 200.0" --filter-name "FS200" \
    -O "results/variants_indels_filtered.vcf"

#    -filter "ReadPosRankSum < -20.0" --filter-name "ReadPosRankSum-20" \
```

```python
%%bash

gatk MergeVcfs \
    -I "results/variants_indels_filtered.vcf" \
    -I "results/variants_snps_filtered.vcf" \
    -O "results/variants_all_filtered.vcf"
```

```python
%%bash

bcftools view -f PASS "results/variants_all_filtered.vcf" -Oz -o "results/variants_all_filtered_PASS.vcf"

bcftools stats -s - "results/variants_all_filtered_PASS.vcf" > "results/variants_all_filtered_PASS.stats"
```

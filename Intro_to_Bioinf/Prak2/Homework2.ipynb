## What factors would make you decide to trim your reads before alignment?

1. The discribution of qualities of terminal nucleotides is mediocre with outliers below 20. Decided to trim it to avoid potential misalignment
2. On GC content plot there is a small peak on 4-6 %GC. Means there is higher percentage of AT-rich sequences that expected. I suggested it could be some junk reads or technical sequences that can be automatically detected and deduplicated/removed. Also it could be polyN sequences so i wanted to remove this as well.

As per qc plot after trimming, this peak wasnt removed so it probably linked to other processed. Nevertheless, I decided to proceed with trimming, as it, as least, removed 7% of duplicates.

## Reference genome name and version

Saccharomyces cerevisiae S288C; NCBI RefSeq assembly GCF_000146045.2; Dec 17, 2014; https://www.ncbi.nlm.nih.gov/datasets/genome/GCF_000146045.2/

```python
%%bash 
bwa index data/GCF_000146045.2_R64_genomic.fna
```

```python
%%bash

bwa mem -t 4 data/GCF_000146045.2_R64_genomic.fna data/SRR19238653_1_trimmed.fastq data/SRR19238653_2_trimmed.fastq 2> results/alignment.bwa.log > results/alignment.sam
```

```python
%%bash
samtools view -b -t 4 results/alignment.sam > results/alignment.bam
```

```python
%%bash

samtools sort -n -t 8 results/alignment.bam > results/alignment.namesorted.bam

# fixmate on the name-sorted file to add mate tags
samtools fixmate -m results/alignment.namesorted.bam results/alignment.fixmate.bam

# Sort by COORDINATES (required for samtools markdup)
samtools sort -t 8 -o results/alignment.positionsorted.bam results/alignment.fixmate.bam

samtools markdup results/alignment.positionsorted.bam results/alignment.marked.bam

```

```python
%%bash

samtools flagstat results/alignment.marked.bam > results/flagstat_report.txt
```

```python
cat results/flagstat_report.txt
```

## Percentage of duplicate reads in the BAM file (after marking duplicates)

```python
%%bash

nreads=$(cat results/flagstat_report.txt | head -n 1| grep -o "[0-9]*" | head -n 1)
          
nduplicated=$(cat results/flagstat_report.txt | head -n 5 | tail -n 1 | grep -o "[0-9]*" | head -n 1)

duplicated_percent=$(echo "scale=10; ($nduplicated / $nreads) * 100" | bc)

         
#echo $nreads
#echo $nduplicated
echo $duplicated_percent
```

## Percentage of reads mapped to the reference after duplicate marking

```python
%%bash

nreads=$(cat results/flagstat_report.txt | head -n 1| grep -o "[0-9]*" | head -n 1)
nmapped=$(cat results/flagstat_report.txt | head -n 7 | tail -n 1 | grep -o "[0-9]*" | head -n 1)          

mapped=$(echo "scale=10; ($nmapped / $nreads) * 100" | bc)

echo $mapped
```
